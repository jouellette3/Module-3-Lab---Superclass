```python
'''
Program: Module 3 Lab.ipynb
Author: Jonathan Ouellette
Date: 11/16/2026
Purpose: This program creates a superclass Vehicle and a subclass Automobile. It prompts the user for information about their car and then displays that information.
'''

#Create a superclass Vehicle
class Vehicle:
    def __init__(self,vehicleType):
        self.vehicleType = vehicleType

#Creates a subclass Automobile that inherits from Vehicle
class Automobile(Vehicle):
    
    #Derives the constructor from the superclass Vehicle and prompts the user for information about their car
    def __init__(self):
        super().__init__("car")

        self.year = input("Enter the year of your car: ")
        self.make = input("Enter the make of your car: ")
        self.model = input("Enter the model of your car: ")
        self.doors = input("How many doors does your car have (2 or 4): ")
        self.roof = input("Enter the roof type (solid or sunroof): ")

    def yourAutomobile(self):
        print(f"Vehicle type: {self.vehicleType}")
        print(f"Year: {self.year}")
        print(f"Make: {self.make}")
        print(f"Model: {self.model}")
        print(f"Number of doors: {self.doors}")
        print(f"Type of roof: {self.roof}")

#Creates class object myCar and calls the yourAutomobile method to display the information about the user's car
myCar = Automobile()
myCar.yourAutomobile()
```
